```python
import requests
from requests.auth import HTTPBasicAuth
user='admin'
password='1234567890'
auth=HTTPBasicAuth(user, password)
headers={"Content-Type": "text/xml"}
url="http://127.0.0.1:3000/contexts.xml"



r=requests.get(url,auth=auth)
print (r.content)
```

```python
from lxml import etree
url="http://127.0.0.1:3000/projects.xml"



r=requests.get(url,auth=auth)
xmltext=r.text.replace('encoding="UTF-8"',"")
xml=etree.fromstring(xmltext)
```

```python
print(etree.tostring(xml,pretty_print=True).decode('UTF8'))
```
